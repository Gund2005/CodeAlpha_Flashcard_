```python
import pandas as pd

# Dataset banavuya
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Clothing', 'Home', 'Electronics', 'Home', 'Electronics'],
    'Sales': [50000, 35000, 60000, 42000, 55000, 65000, 48000, 70000],
    'Profit': [10000, 7000, 12000, 8000, 11000, 13000, 9500, 15000]
}

df = pd.DataFrame(data)
df.to_csv('sales_data.csv', index=False)
print("✅ sales_data.csv file banli!")
print(df)
```

```python
# TASK 2: EDA + TASK 3: Data Visualization
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")

# Data load kara
df = pd.read_csv('sales_data.csv')

print("--- TASK 2: EDA ---")
print("\n1. Dataset Info:")
print(df.info())

print("\n2. Summary Statistics:")
print(df.describe())

print("\n3. Null Values Check:")
print(df.isnull().sum())

print("\n--- TASK 3: Visualization ---")

# Graph 1: Bar Chart - Category nusar Sales
plt.figure(figsize=(10,5))
sns.barplot(x='Category', y='Sales', data=df, estimator=sum, ci=None, palette='viridis')
plt.title('Category nusar Total Sales')
plt.ylabel('Total Sales ₹')
plt.xlabel('Category')
plt.show()

# Graph 2: Line Chart - Month nusar Profit
plt.figure(figsize=(10,5))
sns.lineplot(x='Month', y='Profit', data=df, marker='o', linewidth=2.5, color='red')
plt.title('Mahina nusar Profit Trend')
plt.xticks(rotation=45)
plt.ylabel('Profit ₹')
plt.show()

# Graph 3: Heatmap - Correlation
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Sales vs Profit Correlation')
plt.show()

print("\n✅ EDA + Visualization Complete!")
```
